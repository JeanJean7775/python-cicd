```python
from emir_validator import validate_trade


def test_valid_trade():
    trade = {
        "uti": "UTI-12345",
        "currency": "EUR",
        "notional": 1_000_000
    }

    result = validate_trade(trade)

    assert result["status"] == "PASSED"


def test_missing_uti():
    trade = {
        "uti": None,
        "currency": "EUR",
        "notional": 1_000_000
    }

    result = validate_trade(trade)

    assert result["status"] == "FAILED"
    assert "Missing UTI" in result["errors"]


def test_invalid_notional():
    trade = {
        "uti": "UTI-12345",
        "currency": "EUR",
        "notional": -500
    }

    result = validate_trade(trade)

    assert result["status"] == "FAILED"
    assert "Invalid notional" in result["errors"]
```
